# NumPy: arreglos y matrices aleatorias

## 1. Inspeccionar un arreglo bidimensional

Se convierte una lista de listas en un arreglo de NumPy con datos de tipo flotante. Aunque la variable original se llama `array_1d`, el arreglo tiene **dos dimensiones**: 3 filas y 4 columnas. `type` identifica el objeto; `dtype` identifica el tipo de sus elementos. Se conservan los nombres de las variables del ejemplo original.

```python
import numpy as np
```

```python

lista = [[1, 2, 3, 4],[3, 4, 5, 6], [5, 6, 7, 8]]

array_1d = np.array(lista, dtype='float')


# Tipo de objeto y tipo de sus elementos
print("Tipo de objeto:", type(array_1d))
print("Tipo de elementos (dtype):", array_1d.dtype)

#contenido del array
array_1d
```

```python
# ndim
print('Dimensiones: ', array_1d.ndim)
# shape
print("Shape: ", array_1d.shape)
# size
print("Size: ", array_1d.size)
```

## 2. Generar y mostrar matrices aleatorias

`generar_matriz(N)` crea una lista de `N` filas, cada una con `N` enteros aleatorios entre 1 y 9, ambos incluidos. `imprimir_matriz` organiza esos valores en columnas. Se prueban los tamaños originales de 5 × 5, 10 × 10 y 20 × 20. Los números cambian al volver a ejecutar las celdas porque no se fija una semilla.

```python
import random
```

```python
def generar_matriz(N):
    matriz= []
    for i in range (N):
        fila = []
        for j in range(N):
            fila.append(random.randint(1, 9))
        matriz.append(fila)
    return matriz

N = 5

matriz_aleatoria = generar_matriz(N)
for fila in matriz_aleatoria:
    print(fila)

def imprimir_matriz(matriz):
    for fila in matriz:
        fila_formateada = ' '.join(f'{elemento:2}' for elemento in fila)
        print(fila_formateada)
```

```python
print("Matriz de 5x5")
m_5 = generar_matriz(5)
imprimir_matriz(m_5)
print("============")
print("Matriz de 10x10")
m_10 = generar_matriz(10)
imprimir_matriz(m_10)
print("============")
print("Matriz de 20x20")
m_20 = generar_matriz(20)
imprimir_matriz(m_20)
```

## 3. Explorar filas, columnas y traspuesta

Se consulta la primera fila con `[0]` y la segunda columna con `[:, 1]`. La propiedad `.T` intercambia filas y columnas: un arreglo de forma `(3, 4)` pasa a tener forma `(4, 3)`. Cambia los índices para consultar otras posiciones válidas y vuelve a ejecutar la celda.

```python
print("Primera fila:", array_1d[0])
print("Segunda columna:", array_1d[:, 1])
print("Traspuesta:")
print(array_1d.T)
print("Forma original:", array_1d.shape)
print("Forma de la traspuesta:", array_1d.T.shape)
```

## 4. Probar otros tamaños

Se reutilizan las funciones con matrices de 3 × 3 y 7 × 7. **Modifica `N`** por otro entero positivo y vuelve a ejecutar la celda para observar cómo cambia el tamaño de la matriz. Ejecuta primero las celdas que definen `generar_matriz` e `imprimir_matriz`.

```python
N = 3  # Puedes cambiar N por otro entero positivo.
matriz_3 = generar_matriz(N)
print(f"Matriz de {N}x{N}")
imprimir_matriz(matriz_3)
print("Forma:", (len(matriz_3), len(matriz_3[0])))

print("============")
N = 7  # Segundo tamaño de exploración; también puedes modificarlo.
matriz_7 = generar_matriz(N)
print(f"Matriz de {N}x{N}")
imprimir_matriz(matriz_7)
print("Forma:", (len(matriz_7), len(matriz_7[0])))
```

## Resultados de la exploración

- El arreglo original es un `numpy.ndarray` con elementos `float64`, 2 dimensiones, forma `(3, 4)` y 12 elementos.
- La primera fila contiene `[1., 2., 3., 4.]`; la segunda columna contiene `[2., 4., 6.]`.
- La traspuesta tiene forma `(4, 3)` y conserva los mismos 12 elementos.
- Las matrices generadas son cuadradas y contienen valores entre 1 y 9. Las pruebas adicionales muestran tamaños `(3, 3)` y `(7, 7)` con los valores iniciales de `N`.
- Al cambiar `N` cambia el tamaño; al reejecutar sin cambiarlo pueden cambiar los números aleatorios.
